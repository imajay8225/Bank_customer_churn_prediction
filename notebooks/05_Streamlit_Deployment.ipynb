```

# Save your best trained model as model.pkl first.

streamlit_code = '''
import streamlit as st
import pandas as pd
import joblib

model=joblib.load("model.pkl")

st.title("Bank Customer Churn Prediction")

uploaded=st.file_uploader("Upload CSV")

if uploaded:
    df=pd.read_csv(uploaded)
    pred=model.predict(df)
    prob=model.predict_proba(df)[:,1]
    df["Prediction"]=pred
    df["Probability"]=prob
    st.dataframe(df)
    st.download_button("Download",df.to_csv(index=False),"prediction.csv")
'''

print(streamlit_code)
```
